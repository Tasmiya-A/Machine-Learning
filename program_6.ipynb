```python
import random
def uniform_random():
  return random.uniform(0,1)
def f(x):
  return x**2
N=10
samples=[uniform_random() for _ in range(N)]
print("samples:",samples)
f_values=[f(x) for x in samples]
print("f_values:",f_values)
I_est=sum(f_values)/N
print("I_est:",I_est)
I_true=1/3
print("True_value:",I_true)
```
